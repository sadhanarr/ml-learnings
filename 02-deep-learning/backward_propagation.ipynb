```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define Inputs for the XOR problem
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]]).T
y = np.array([0,1,1,0])
```

## Initialize network parameters

Define network parameters
*   Assign the number of input layers
*   Number of hidden layers
*   Number of output layers
*   Learning rate
*   Number of epochs
*   Initialize the weigths and bias





```python
def initialize_network_parameters():
  inputSize = 2
  hiddenSize = 2
  outputSize = 1
  epochs = 180000
  learningRate = 0.1

  #Intialize weights and biases to the nodes
  # The weight matrix is organized such that each row corresponds to the neurons in the next layer and each column corresponds to the neurons in the current layer.
  # This makes matrix multiplication straightforward during the forward pass.
  # The bias matrix is a column vector with each row corresponding to a neuron in the respective layer. This allows each neuron to have its own bias term.

  w1 = np.random.rand(hiddenSize,inputSize) * 2 - 1 # weights from input to hidden layer
  b1 = np.random.rand(hiddenSize,1) * 2 -1 # bias to the hidden layer
  w2 = np.random.rand(outputSize,hiddenSize) * 2 - 1 # weights from hidden to output layer
  b2 = np.random.rand(outputSize,1) * 2 -1 # bias to the output layer

  return w1,b1,w2,b2,epochs,learningRate

```

## Train the neural network

The training through neural network happens in 5 steps;

1. **Forward Pass**

* The input X is multiplied by the weights of the first layer and added to the bias to compute the activations of the hidden layer.
* The activations of the hidden layer are then multiplied by the weights of the second layer and added to the bias to compute the output of the network.

2. **Error calculation**

*   The error is calculated as difference between (y) and the output (a2)

3. **Backward Pass**

*   **Output Layer** - The derivative of the sigmoid function is applied to the error, producing gradient for the output(da2). This is used to calculate how much weights in the output layer need to be adjusted.
*   **Hidden Layer** - The error is then propagated backwards to the hidden layer. The gradient is computed here by taking the dot product of Transpose of weights and gradient from output layer. The derivative of the activation function is used to adjust this error

4. **Update weights and bias**

*   After computing the gradients, updating the weights and bias happen through learning rate and gradients

5. **Training**

* This entire process is repeated for many iterations (epochs). During each epoch, the model adjusts the weights and bias to reduce the error.


```python
w1,b1,w2,b2,epochs,lr = initialize_network_parameters()
error_list = []
for epoch in range(epochs):
  #Forward Pass
  z1 = np.dot(w1,X) + b1 # weighted sum for hidden layer
  a1 = 1/(1+np.exp(-z1)) #Sigmoid Activation Function

  z2 = np.dot(w2,a1) + b2 # weighted sum for output layer
  a2 = 1/(1+np.exp(-z2)) # Activation function of output layer

  #Error calculation
  error = y - a2

  #Back propagation

  # Applying Chain Rule for derivative : dE/dz2 = dE/da2 * da2/dz2
  # dE/da2 = d[1/2*(d-a2)^2]/da2 = -(d - a2) = -error
  # da2/dz2 = a2 * (1 - a2)  (derivative of the sigmoid function)
  # => dE/dz2 = -error * a2 * (1 - a2)
  # We use the negative of this gradient (i.e. error * a2*(1-a2)) so that
  # updating weights with w += lr * dw moves them in the direction that
  # reduces the error (gradient descent).
  da2 = error * (a2 * (1-a2))
  dz2 = da2

  da1 = np.dot(w2.T,dz2)
  dz1 = da1 * (a1 * (1-a1))

  #Update weights and bias
  w2 += lr * np.dot(dz2,a1.T)
  b2 += lr * np.sum(dz2,axis=1,keepdims=True)

  w1 += lr * np.dot(dz1, X.T)  # Update weights from input to hidden layer
  b1 += lr * np.sum(dz1, axis=1, keepdims=True)  # Update bias for hidden layer
  if (epoch+1)%10000 == 0:
      print("Epoch: %d, Average error: %0.05f"%(epoch, np.average(abs(error))))
      error_list.append(np.average(abs(error)))
```

## Testing the Network
 Compute the final value of a2 after training.
 Observed that they are close to the ground truth with minimal error

```python
# Testing the trained network
z1 = np.dot(w1, X) + b1  # Weighted sum for hidden layer
a1 = 1 / (1 + np.exp(-z1))  # Sigmoid activation for hidden layer

z2 = np.dot(w2, a1) + b2  # Weighted sum for output layer
a2 = 1 / (1 + np.exp(-z2))  # Sigmoid activation for output layer

# Print results
print('Final output after training:', a2)
print('Ground truth', y)
print('Error after training:', error)
print('Average error: %0.05f'%np.average(abs(error)))
```
